```python
import re

import random
```

```python
class RailwayReservationSystem:

    def __init__(self):
        self.train_name = "Rajdhani Express"
        self.available_seats = 5
        self.passengers = []

    def book_status(self,name , pnr , seat , fare):

        pasengers = {
        "name" : name,
        "pnr" : pnr,
        "seat" : seat,
        "fare" : fare
        }
        self.passengers.append(pasengers)

    def validate_ticket(self, pnr , seat):
        pnrr = r"^\d{8}$"
        sett = r"^[A-Z][0-9]{1,2}$"

        pn_va = re.fullmatch(pnrr , pnr)
        seat_va = re.fullmatch(sett , seat)

        return pn_va and seat_va

    def alloted_seat(self):
        if self.available_seats > 0 :
            self.available_seats -= 1
            return self.available_seats
            return "Seat Allocated"

        return " No Seats Available"  
```

```python
wer = RailwayReservationSystem()
wer
```

```python
wer.book_status("SNEH" , "89891234" , "S4" , 1300)
```

```python
for h in wer.passengers:
    print(h)
```

```python
for passenger in wer.passengers:
    print(passenger["name"], "-" , wer.alloted_seat())
```
